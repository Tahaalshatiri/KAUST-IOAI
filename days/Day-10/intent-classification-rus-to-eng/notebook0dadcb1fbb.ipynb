```python
# Fast Sanity Baseline — Majority Intent + O Slots
#
# This is not meant to score well. It verifies the submission format.

from pathlib import Path
import pandas as pd

DATASET_PATH = Path("/kaggle/input/YOUR_CROSSLINGUAL_NLU_DATASET")

sample = pd.read_csv(DATASET_PATH / "sample_submission.csv")
train_text = (DATASET_PATH / "train.conll").read_text(encoding="utf-8")

# Naive extraction of first intent frequency from token rows.
intents = []
for line in train_text.splitlines():
    if not line or line.startswith("#"):
        continue
    parts = line.split("\t") if "\t" in line else line.split()
    if len(parts) >= 3:
        intents.append(parts[2])
majority_intent = pd.Series(intents).value_counts().index[0]

sub = sample.copy()
sub["intent"] = majority_intent
# sample already contains the right number of O tags per row.
sub.to_csv("/kaggle/working/submission.csv", index=False)
print(sub.head())
print("Saved /kaggle/working/submission.csv")
```
